#**Logistic Regression: model in PyTorch**

```python
import torch
from torch import nn

class TorchLogReg(nn.Module):                                  #NN constructor
    def __init__(self, input_dim, num_classes):               #with number of input and output
        super().__init__()
        self.linear = nn.Linear(input_dim, num_classes)

    def forward(self, x):
      return self.linear(x)
```

```python
torch_log_reg = TorchLogReg(784, 10).cuda()                          #initialize NN with 784 input + 1 bias + 10 output
                                                                     #.cuda(): choses the invidia gpu

loss = nn.CrossEntropyLoss(reduction = 'sum')                        #sum(y_i log(p_i)): classic cross entropy loss
#loss = nn.MSELoss(reduction = 'sum')                                #mean square error loss

optimizer = torch.optim.SGD(torch_log_reg.parameters(), lr=0.01)
X_train = X_train.to('cuda')
y_train = y_train.to('cuda')
X_valid = X_valid.to('cuda')
y_valid = y_valid.to('cuda')

for epoch in range(1000):

  #OPTIMIZATION PHASE FOR EACH EPOCH
  optimizer.zero_grad()                                     #weights initialisation to None:                       WEIGHTS INITIALISATION
  y_pred = torch_log_reg(X_train)                           #forward phase on training data, output y_predicted:   PREDICT OUTPUT

  loss_result = loss(y_pred, torch.tensor(y_train))         #compute the loss between y_predicted and y_train:     LOSS
  #- torch.tensor(y_train) : wraps y_trian into a PyTorch tensor, to compare it to y_pred which is already a PyTorch tensor
  #- the best/appropriate activation function is already included into the loss

  loss_result.backward()                                    #backward phase, to compute the gradient:              GRADIENT
  optimizer.step()                                          #applies the -learning rate x gradient to the weights: OPTIMISATION


  if epoch % 50 == 0:
    with torch.no_grad():
      pred_valid = torch_log_reg(X_valid)
    prediction = torch.argmax(pred_valid.to('cpu'), dim=1)
    accuracy = accuracy_score(y_valid.to('cpu'), prediction)
    print(f"Accuracy is {accuracy*100}%")
```
