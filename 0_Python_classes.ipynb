# 0) Python Classes

**CLASS**: FRAMEWORK TO CREATE OBJECTS with attributes, AND DO ACTIONS/APPLY METHODS TO THEM in form of functions


**CONVENTION**: start classes with capital letter
              : call objects by small letter


RMK) methods/functions defined inside a class hold only to the objects of that class


## Define a class with attributes and functions

```python
#1) CREATE AN OBJECT WITH ATTRIBUTES
#standard syntax:

#class ClassName:
#    def __init__(self, attribute1, attribute2):     #refer to the object as self in the class
#        self.attribute1 = attribute1
#        self.attribute2 = attribute2


class Dog:
    def __init__(self, name):        #initialise a new object with attribute name
        self.n = name       

class Dog1:
    def __init__(self, name, age):   #initialise a new object with attributes name and age
        self.n = name
        self.a = age


mydog = Dog1("Jack", 10)    #creates the new object mydog with the given attributes

mydog.n       #access the attribute name of the object mydog
```

```python
#2) METHODS/ACTIONS IN THE CLASS to apply to the objects defined in the class
#== functions inside the class, referring to the object as "self"


class Dog2:
    def __init__(self, name, age):    #1) initialisation of the object with attributes name and age
        self.n = name
        self.a = age

    def bark(self):                   #2) method to apply to the object
        print(f"Woof! My name is {self.n} and I am {self.a} years old.")



mydog2 = Dog2("Jack", 10)    #creates the new object mydog with the given attributes
mydog2.bark()                #call the function/method bark to the object mydog2
```

### Example: Rectangle class with methods to compute perimeter and area

```python

class Rectangle:
    def __init__(self, height, width):        #1) object initialisation with sides
        self.a = height
        self.b = width

    def perimeter(self):             #2) method to compute perimeter
        return 2*(self.a + self.b)

    def area(self):                  #2) method to compute area
        return self.a * self.b



rec = Rectangle(5, 10)                    #new object rectangle created with attributes sides of value 5 and 10
print('Perimeter is', rec.perimeter())    #call method perimeter to the object rec
print('Area is', rec.area())              #call method area to the object rec
```

## Subclasses and inheritance

Starting from a parent class, we can create a child class that inherits the attributes and methods of the parent class.

It is possible to
- keep the methods of the parents class
- redefine them with super().method()
- add new methods to the child class

```python

#STANDARD SYNTAX:
#class ChildClass(ParentClass):
#    def __init__(self, attribute1, attribute2):
#        super().__init__(attribute1, attribute2)         #super() calls the parent class constructor
#    def method1(self):
#        super().method1()                                #super() calls the method of the parent class.....

```

### Example: square class from rectangle

```python

class Square(Rectangle):
    def __init__(self, a):           #initialise the subclass
        super().__init__(a, a)       #using the parent class constructor with attributes a, a

    def circle_inside_square(self):    #new method to the subclass
        return 3.14 * (self.a/2)**2


sq = Square(5)                                               #create object square with side of value 5

                                                             #the following functions are inherited from the parent class Rectangle
print(f"The perimeter of the square is {sq.perimeter()}")    #call method perimeter to the object sq
print(f"The area of the square is {sq.area()}")              #call method area to the object sq

print(f"The area of the circle inside the square is {sq.circle_inside_square()}")     #call method circle_inside_square to the object sq
```

## isistance(object, ClassName)

Good practice: check if the type of object is that of the class ClassName.

```python

isinstance(sq, Square)             #check if sq is an instance of the class Square
```

```python
isinstance(sq, Rectangle)          #check if sq is an instance of the class Rectangle
```
