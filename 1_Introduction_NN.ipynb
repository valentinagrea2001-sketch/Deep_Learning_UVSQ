## Basic definitions of Deep Forward NNs == MLPs

>*A deep feedforward neural network (NNs) is also referred to as multilayer perceptron (MLPs)*.
>
>**Deep feedforward NNs == MLPs** consist of *fully connected*:
>
>- *input layer*: receiving raw data and passing them to the first hidden layer;
>- *hidden layers*: intermediate layers that transform the information through a linear function and apply then to them a non-linear activation function;
>- *output layer*: produces the final output.

>A **deep feedforward neural network == MLP** is a composition of functions or layers $f^{(n)}(....f^{(2)}(f^{(1)}(x)))$, applied to a vector of input data $x \in \mathbb{R}^{p}$ and returning a vector of output data $y \in \mathbb{R}^{m}$.
>
>- The *input layer* contains the raw data $x\in \mathbb{R}^{p}$ and passes them to the first hidden layer as a linear combination.
>
>- The *hidden layers* are expressed as the composition of an activation function $\sigma$ and a linear function, as $f^{\left(i\right)}: \mathbb{R}^{d}\rightarrow \mathbb{R}^{d'}$:
>$$
f^{(i)}(x) = \sigma( W_{i} x + b_{i} ).
>$$
>They receive the input data from the previous layer, weight them through a matrix $W_{i}$ and add a bias term $b_{i}$ (linear function). Then, a non-linear activation function $\sigma$ is applied, in order to solve problems where data are not linearly separable.
>
>- The *output layer* returns the ouput $y\in\mathbb{R}^{m}$ as a linear combination. This layer can eventually contain a non-linear activation function $\sigma$ depending on the task.
>

> Briefly, a MLP is:
>$$
>f(x)=
>\sigma^{(n)}
>\left(
>W^{(n)}
>\sigma^{(n-1)}
>\left(
>\cdots
>\sigma^{(1)}(W^{(1)}x+b^{(1)})
>\cdots
>\right)
>+b^{(n)}
>\right)
$$

**A neural network is defined by**:

- number of input units in the input layer;

- number of output units in the output layer;

- number of hidden (intermediate) layers and number of units for each;

- weights and bias for each layer;

- activation function.

As the name says, neural networks are usually *graphically* represented as a network of units and arrows from a layer to the next one.

**Remarks**

- NNs are assumed to be *Fully connected*, i.e. units of two subsequent layers are fully connected.

- NNs are *feedforward*, because the information flows from the input layer, through the hidden layers, until the output layer. Equivalently, the network takes in input some data, processes them sequentially through the hidden layers, until it produces an output.

- the same activation function is applied to ALL hidden layers, and eventually to the ouput layer depending on the task.

- the activation function applied to the output layer can be different, depending on the task.

## The activation functions

**The need of non-linear activation functions to solve problems with non-linearly separable data**:

The composition of linear functions is a linear function. Therefore, if all hidden layers would be just linear, our neural network would be only able to work with linearly separable data.

To add non-linearity and allow for solving non-linear and more complex tasks, between the hidden/intermediate layers is usually applied a non-linear function called **activation function**. Sometimes, depending on the task, this activation function is applied to the output layer.

***Most common activation functions, to add non-linearity to the model / hidden layers:***

**$sign(x)$** :
- commonly used at the very beginning of neural networks, in the 60s;
- binary classification tasks.

**$\sigma(x)=\dfrac{1}{1+e^{-x}}$** :
- alternative to $sign(x)$;
- binary classification tasks;
- maps in $[0,1]$; 
- PROS: smooth and differentiable, which is helpful for gradient computations and weights optimization; no information is lost;
- CONS: can yield to vanishing gradient.

**$ReLu(x)=max(0,x)$** : 
- favourite in CNNs and other MLPs;
- maps in $[0, +\infty)$;
- PROS: computationally cheap, because derivatives are always 0 or 1 for gradient computation;
- CONS: not differentiable in 0, yields to lost of information.


## Frameworks for tabular data

Tabular data are the most common kind of data, the ones where today feedforward NNs == MLPs are having success.

**Frameworks to work with tabular data:**

*bootstrap methods*: such as random forest:
- fast and easy to implement;
- enough precise.

*AutoML*: automatic implementation of MLPs with the best choices:
- it does everything by itself;
- difficult to be used.

*TablCL*: foundational model of INRIA Paris.

*TabPFN*: based on transformers.

*sklearn MLP*: sklearn on MLPs:
- work only on CPUs, slow;
- low precision.

*RealMLP*: modern implementation of MLPs:
- can work on GPUs with command "device='cuda'", faster; 
- efficient optimisation methods;
- many activation functions;
- data normalisation between layers;
- many activation functions.
