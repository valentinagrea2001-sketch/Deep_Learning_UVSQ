## Basic definitions of Deep Forward NNs == MLPs

>*A deep feedforward neural network (NNs) is also referred to as multilayer perceptron (MLPs)*.
>
>**Deep feedforward NNs == MLPs** consist of *fully connected*:
>
>- *input layer*: receiving raw data and passing them to the first hidden layer;
>- *hidden layers*: intermediate layers that transform the information through a linear function and apply then to them a non-linear activation function;
>- *output layer*: produces the final output.

>A **deep feedforward neural network == MLP** is a composition of functions or layers $f^{(n)}(....f^{(2)}(f^{(1)}(X)))$, applied to a matrix of raw data $X \in \mathbb{R}^{n\times p}$ and returning a matrix or vector of output data $Y$.
>
>- The *input layer* contains the raw data $X\in \mathbb{R}^{n\times p}$ and passes them to the first hidden layer as a linear combination.
>
>- The *hidden layers* are expressed as the composition of a linear function and an activation function $\sigma: \mathbb{R}^{k} \rightarrow \mathbb{R}^{k}$, as $f^{\left(i\right)}: \mathbb{R}^{n\times h_{i-1}}\rightarrow \mathbb{R}^{n\times h_{i}}$:
>$$
f^{(i)}(X^{(i-1)}) = \sigma( W_{i} \cdot X^{(i-1)} + b_{i} ),
>$$
>where $X^{(i-1)} \in \mathbb{R}^{n\times h_{i-1}}$ is the output data from the previous $i-1$ layer, $W_{i}\in\mathbb{R}^{h_{i-1}\times h_{i}}$ is the weights matrix, $b_{i}\in\mathbb{R}^{h_{i}}$ is the bias term, and $\sigma$ is the activation function. Observe that the activation function is applied *element by element*.
>
>An intermediate hidden layer receives the input data from the previous layer, weights it through a matrix $W_{i}$ and adds a bias term $b_{i}$ (linear function until here). Then, a non-linear activation function $\sigma$ is applied element by element. The latter has the advantage of making the network able to solve problems where data are not linearly separable.
>
>- The *output layer* returns the ouput matrix or vector $Y$ as a linear combination. This layer can eventually contain a non-linear activation function $\sigma'$ depending on the task.
>

> Briefly, a MLP is:
>$$
>f(X)=
>\sigma^{(n)}
>\left(
>W^{(n)}
>\sigma^{(n-1)}
>\left(
>\cdots
>\sigma^{(1)}(W^{(1)}\cdot X+b^{(1)})
>\cdots
>\right)
>+b^{(n)}
>\right)
$$

**To define a Neural Network we need to set**:

- number of hidden/intermediate layers (*depth*) and number of units for each layer;

- number of input units in the input layer;

- number of output units in the output layer;

- activation function for the hidden layers and the one eventually for the output layer.

As the name says, neural networks are usually *graphically* represented as a network of units and arrows from a layer to the next one.

**In a neural network we need to learn**:

- weights and biases.

**Remarks**

- NNs are assumed to be *Fully connected*, i.e. units of two subsequent layers are fully connected.

- NNs are *feedforward*, because the information flows from the input layer, through the hidden layers, until the output layer. Equivalently, the network takes in input some data, processes them sequentially through the hidden layers, until it produces an output.

- the *same activation function is applied to ALL hidden layers and ELEMENT BY ELEMENT in each layer*. It is eventually applied to the ouput layer depending on the task. In the latter case, the activation function applied to the output layer can be different from the one applied in hidden layers, depending on the task.

- our definition was stated for input matrices, but actually it works for vectors also.

- the linear function firstly applied to layers is also called affine applications.

## The activation functions

**The need of non-linear activation functions to solve problems with non-linearly separable data**:

The composition of linear functions is a linear function. Therefore, if all hidden layers would be just linear, our neural network would be only able to work with linearly separable data.

To add non-linearity and allow for solving non-linear and more complex tasks, between the hidden/intermediate layers is usually applied a non-linear function called **activation function**. Sometimes, depending on the task, this activation function is applied to the output layer.

***Most common activation functions, to add non-linearity to the model / hidden layers:***

**$sign(x)$** : sign
- commonly used at the very beginning of neural networks, in the 60s;
- binary classification tasks.
- CONS: its derivative is zero everywhere except at the origin, where it is undefined. This causes problems during weights optimisation phase, lost of information and vanishing gradient problem.

**$\sigma(x)=\dfrac{1}{1+e^{-x}}$** : sigmoid
- alternative to $sign(x)$;
- binary classification tasks;
- maps in $[0,1]$; because of the specific shape of this function around $0$, data are naturally classified as $0$ or $1$.; 
- PROS: smooth and differentiable, which is helpful for gradient computations and weights optimization; no information is lost;
- CONS: can yield to vanishing gradient.

**$\sigma(z_0,\dots,z_{n})=\dfrac{1}{\sum_{i=0}^{n} e^{z_i}}\cdot \left(e^{z_0},\dots,e^{z_{n}}\right) \in\mathbb{R}^{n}$**: softmax
- generalisation of sigmoid, for n=1 we find again the sigmoid;
- multiclass classification in $[|0,n|]$ classes.

**$ReLu(x)=max(0,x)$** : ReLu
- favourite in CNNs and other MLPs;
- maps in $[0, +\infty)$;
- PROS: computationally cheap, because derivatives are always 0 or 1 for gradient computation;
- CONS: not differentiable in 0, yields to lost of information.


Rmk) *derivatives 0 == problems during optimisation phase == lost of information == vanishing gradient problem*

## Frameworks for tabular data

Tabular data are the most common kind of data, the ones where today feedforward NNs == MLPs are having success.

**Frameworks to work with tabular data:**

*bootstrap methods*: such as random forest:
- fast and easy to implement;
- enough precise.

*AutoML*: automatic implementation of MLPs with the best choices:
- it does everything by itself;
- difficult to be used.

*TablCL*: foundational model of INRIA Paris.

*TabPFN*: based on transformers.

*sklearn MLP*: sklearn on MLPs:
- work only on CPUs, slow;
- low precision.

*RealMLP*: modern implementation of MLPs:
- can work on GPUs with command "device='cuda'", faster; 
- efficient optimisation methods;
- many activation functions;
- data normalisation between layers;
- many activation functions.


> We will work with MLPs on **RealMLP**, selecting *" device='cuda' " to employ GPUs*.
