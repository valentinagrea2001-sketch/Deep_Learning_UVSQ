# Logistic Regression with PyTorch

```python
#NN CREATION

torch_log_reg = TorchLogReg(784, 10)                    #input 784, output 10, bias 1
```

```python
#LOSS FUNCTION

loss = nn.CrossEntropyLoss(reduction='sum')
```

```python
#OPTIMISATION

#GRADIENT DESCENT: -(gradient) is added to the weights at each iteration of the algorithm, to proceed towards the minimum of the empirical risk

#STOCHASTIC GRADIENT RESEARCH (SDG): to prefer because does not stuck into minima

optimizer = torch.optim.SGD(torch_log_reg.parameters(), lr=0.01)              #lr: learning rate
```

```python
#EPOCHS AND MODEL'S EVALUATION

#Always define an epoch, i.e. the number of iterations to run the algorithm

for epoch in range(1000):

  #OPTIMIZATION PHASE FOR EACH EPOCH
  optimizer.zero_grad()                                     #weights initialisation to None:                       WEIGHTS INITIALISATION
  y_pred = torch_log_reg(train_torch)                       #forward phase on training data, output y_predicted:   PREDICT OUTPUT
  loss_result = loss(y_pred, torch.tensor(y_train))         #compute the loss between y_predicted and y_train:     LOSS
  loss_result.backward()                                    #backward phase, to compute the gradient:              GRADIENT
  optimizer.step()                                          #applies the -learning rate x gradient to the weights: OPTIMISATION

 #MODEL'S EVALUATION FOR EACH EPOCH
  if epoch % 50 == 0:                                       #every 50 epochs, evaluate the model
    with torch.no_grad():
      pred_valid = torch_log_reg(X_valid_torch)
    prediction = torch.argmax(pred_valid, dim=1)
    accuracy = accuracy_score(y_valid, prediction)
    print(f'Accuracy: {accuracy}')
```
