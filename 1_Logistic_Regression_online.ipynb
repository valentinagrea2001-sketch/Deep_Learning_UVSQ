***Activation functions** to add non-linearity to the model.*

Indeed, already the intermediate arrows are linear. A linear activation function would make the model to solve only linear tasks.

Examples of activation functions:

- $sign(x)$ : commonly used in 60s at the very beginning for binary classification tasks.

- $\sigma(x)=\dfrac{1}{1+e^{-x}}$   : sigmoid function. It was born for binary classification tasks, at the place of $sign(x)$, because it's continuous and differentiable. Still a lot chosen for binary classification tasks.

- $ReLu(x)=max(0,x)$ : derivatives are always 0 or 1, very easy for the backward phase.

- $tanh(x)$
