# 2) Activation functions plots

> Determine how a neuron transforms its input before passing it to the next layer.

> Introduce non-linearity, which allows NN to learn complex patterns instead of just linear relationships.

```python
from matplotlib import pyplot as plt

import torch
from torch import nn
```

```python
# X-AXIS

x = torch.arange(-10, 10, 0.1)
```

## Sigmoid

- Binary classification task $0, 1$

- Maps in $[0,1]$

- PROS : smooth and differentiable

- CONS : can yield to vanishing gradient

```python
act = nn.Sigmoid()

plt.plot(x, act(x))

plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```

## Tanh

- Used in RNNs, zero-centered representations

- Maps in $[-1, 1]$, zero-centered

- PROS : smooth and differentiable

- CONS : can yield to vanishing gradient

```python
act = nn.Tanh()

plt.plot(x, act(x))


plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```

## ReLu

- Hidden layers in CNNs and MLPs (multi-layers perceptron)

- $ReLu(x) = max(0, x)$, maps in $[0, +\infty]$

- PROS : computationally cheap, derivatives are 1 or 0

- CONS : not differentiable at 0, dying neurons

```python
act = nn.ReLU()

plt.plot(x, act(x))

plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```

## Leaky ReLu

- Hidden layers in CNNs, MLPs, and models where ReLU neurons become inactive

- $LeakyReLu(x)=max(x, \alpha x)$

- PROS : computationally cheap, nonzero gradient on negative inputs (unlike ReLu)

- CONS : negative slope is usually fixed and must be chosen, it does not guarantee that dying neurons will be eliminate

```python
act = nn.LeakyReLU(0.05)                 # usually alpha=0.01

plt.plot(x, act(x))

plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```

## SeLu

- Specially configured deep feedforward NN designed to be self-normalizing

PROS : self-normalizing.

CONS : self-normalization depends on assumptions about the network, initialization, and data distribution; it is not guaranteed.

```python
act = nn.SELU()

plt.plot(x, act(x))

plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```

## GeLu

- Transformers and modern deep learning models

- PROS : weights inputs are according to gaussian distribution, does not set to 0 all negative values as ReLu

- CONS : more computationally expensive than ReLu

```python
act = nn.GELU()

plt.plot(x, act(x))

plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```
