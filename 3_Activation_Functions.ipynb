# Activation functions plots

```python
from matplotlib import pyplot as plt

import torch
from torch import nn
```

```python
# X-AXIS

x = torch.arange(-10, 10, 0.1)
```

## ReLu

```python
act = nn.ReLU()

plt.plot(x, act(x))

plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```

## SeLu

```python
act = nn.SELU()

plt.plot(x, act(x))

plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```

## GeLu

```python
act = nn.GELU()

plt.plot(x, act(x))

plt.xticks(torch.arange(-10, 11, 2))
plt.grid(True)
```
