# 4) Multi-Layer Perceptron VS RealMLP

> **MLP**: Multi Layer Perceptron

> **RealMLP**: Multi Layer Perceptron with automatic preprocessing of tabular data

This notebook will compare a bootstrap method, the classic MLP, and the RealMLP, for the multiclass classification task.

In this repository, RealMLP will not be used again, it's just an example about how it works.

```python
import openml

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
task = openml.tasks.get_task(361113)

dataset = openml.datasets.get_dataset(task.dataset_id, download_data=False)

X, y, categorical_indicator, attribute_names = dataset.get_data(
    dataset_format='dataframe',
    target=task.target_name,
)

X = X.astype(float).values
y = y.astype(int).values
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=15000, random_state=0
)

X_test = X_test[:10000]             # keep 10000 first data
y_test = y_test[:10000]
```

## Bootstrap approach, Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier()
```

```python
%%time

model.fit(X_train, y_train)
```

```python
accuracy_score(y_test, model.predict(X_test))
```

## Classic MLP

```python
from sklearn.neural_network import MLPClassifier
```

```python
# IMPORT classic MLP

model = MLPClassifier(hidden_layer_sizes=(100, 200, 100))
```

```python
# TRAINING
# %%time

model.fit(X_train, y_train)
```

```python
accuracy_score(y_test, model.predict(X_test))
```

```python
# save file

from pathlib import Path
import shutil

repo = Path("realmlp-td-s_standalone")

if not repo.exists():
    raise FileNotFoundError(
        f"Repository not found: {repo.resolve()}"
    )

files = list(repo.glob("*.py"))

if not files:
    print("No Python files found in:", repo.resolve())
else:
    for file in files:
        destination = Path.cwd() / file.name

        if destination.exists():
            print(f"Skipped (already exists): {file.name}")
        else:
            shutil.move(str(file), str(destination))
            print(f"Moved: {file.name}")

    print("Finished.")
```

```python
# read files
 
import re

for fname in ['mlp.py', 'preprocessing.py']:

    p = Path(fname)
    text = p.read_text()
    text = re.sub(r'np\.NaN', 'np.nan', text)
    text = re.sub(r'np\.Inf', 'np.inf', text)
    p.write_text(text)
```

```python
# import RealMLP

from importlib import reload

import mlp

reload(mlp)

from mlp import Standalone_RealMLP_TD_S_Classifier
```

```python
# TRAIN MLP

%%time

realmlp = Standalone_RealMLP_TD_S_Classifier()

realmlp.fit(X_train, y_train)
```

```python
accuracy_score(y_test, realmlp.predict(X_test))
```
