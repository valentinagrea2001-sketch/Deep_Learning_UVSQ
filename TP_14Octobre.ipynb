```python
import openml
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
task = openml.tasks.get_task(361113)

dataset = openml.datasets.get_dataset(task.dataset_id, download_data=False)
X, y, categorical_indicator, attribute_names = dataset.get_data(
    dataset_format='dataframe',
    target=task.target_name,
)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=15000, random_state=0
)
X_test = X_test[:10000]
y_test = y_test[:10000]

```

```python
X_train
```

```python
y_train
```

```python
from sklearn.neural_network import MLPClassifier
```

```python
X_train = X_train.astype(float).values
X_test = X_test.astype(float).values
y_train = y_train.astype(float).values
y_test = y_test.astype(float).values
```

```python
model = MLPClassifier(hidden_layer_sizes=(100, 200, 100))

model.fit(X_train, y_train)
```

```python
accuracy_score(y_test, model.predict(X_test))
```
